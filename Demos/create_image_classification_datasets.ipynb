```python
from multiversity.multicare_dataset import MedicalDatasetCreator
```

```python
mdc = MedicalDatasetCreator(directory = 'medical_datasets')
```

Double-check supported keywords for a disease

```python
search_term = ('malaria')
keywords = [kw for kw in sorted(mdc.keyword_list) if search_term in kw.lower()]
print(keywords)
```

```python
def filter_generator(keywords):
    return  [
        {'field': 'keywords', 'string_list': keywords, 'match_type': 'partial_match'},
        {'field': 'case_strings', 'string_list': keywords},
        {'field': 'label', 'string_list': ['chart'], 'operator': 'none'},
    ]
```

The 7 diseases: **LEPTOSPIROSIS** (VÀNG DA), **MELIOIDOSIS** (WHITMORE), **MALARIA** (SỐT RÉT), **NEUROCYSTICERCOSIS** (SÁN LỢN), **DENGUE**, **SCRUB TYPHUS** (SỐT MÒ)

```python
LEPTOSPIROSIS_KEYWORDS = [
    'leptospirosis',
]

leptospirosis_filter = filter_generator(LEPTOSPIROSIS_KEYWORDS)

mdc.create_dataset(dataset_name='leptospirosis', filter_list=leptospirosis_filter, dataset_type='multimodal')
```

```python
MELIOIDOSIS_KEYWORDS = [
    'melioidosis'
]

melioidosis_filter = [
    {'field': 'keywords', 'string_list': MELIOIDOSIS_KEYWORDS},
    {'field': 'case_strings', 'string_list': ['melioidosis', 'whitmore'], 'operator': 'any'},
    {'field': 'label', 'string_list': ['chart'], 'operator': 'none'}
]

mdc.create_dataset(dataset_name='melioidosis', filter_list=melioidosis_filter, dataset_type='multimodal')
```

```python
dengue_keywords = ['dengue']

dengue_filter = filter_generator(dengue_keywords)

mdc.create_dataset(dataset_name='dengue', filter_list=dengue_filter, dataset_type='multimodal')
```

```python
typhoid_keywords = ['typhoid', 'typhoid fever', 'typhoid intestinal perforation', 'typhoid retinopathy']

typhoid_filter = [
    {'field': 'keywords', 'string_list': typhoid_keywords, 'operator': 'any', 'match_type': 'full_match'},
    {
        'field': 'case_strings',
        'string_list': ['typhoid', 'typhoid fever', 'typhoid intestinal perforation', 'typhoid retinopathy'],
        'operator': 'any',
        'match_type': 'full_match'
    }
]

mdc.create_dataset(dataset_name='typhoid', filter_list=typhoid_filter, dataset_type='multimodal')
```
